```python
import json
from pathlib import Path
import pandas as pd

# 1. Define Paths relative to the notebooks folder location
PROJECT_ROOT = Path("..").resolve()
DATA_DIR = PROJECT_ROOT / "data"

# Aligned perfectly to your structural directory outputs
CHUNKS_PATH = DATA_DIR / "chunks" / "chunks.json"
SETUP_SUMMARY_PATH = DATA_DIR / "processed" / "retrieval_setup_summary.json"
VECTOR_DB_DIR = PROJECT_ROOT / "storage" / "vector_db"

# 2. Resilient JSON Extraction Function
def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

# 3. Execute Loading Sequence
chunks = load_json(CHUNKS_PATH)
setup_summary = load_json(SETUP_SUMMARY_PATH)

# 4. View Statistics
print("=== PIPELINE ENVIRONMENT METRICS ===")
print(f"Chunks Loaded:        {len(chunks)}")
print(f"Vector DB Exists:     {VECTOR_DB_DIR.exists()}")
print(f"Collection Processed: {setup_summary.get('collection_name')}")
print(f"Status Flag:          {setup_summary.get('status')}")

# Output the summary dictionary object at the end of the cell
setup_summary
```

```python
chunk_df = pd.DataFrame(chunks)
chunk_df.head()
```

```python
#Chunk length stats
chunk_df["char_length"] = chunk_df["text"].astype(str).apply(len)

pd.DataFrame([{
    "chunks": len(chunk_df),
    "avg_char_length": round(chunk_df["char_length"].mean(), 2),
    "median_char_length": round(chunk_df["char_length"].median(), 2),
    "min_char_length": int(chunk_df["char_length"].min()),
    "max_char_length": int(chunk_df["char_length"].max()),
}])
```

```python
#Chunk distribution by source file
chunk_df.groupby("source_file").size().sort_values(ascending=False).head(20)
```
